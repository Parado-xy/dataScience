```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        hash_map = dict()
        count = 0
        hash_map[f'{count}'] = ''
        for i, value in enumerate(s):
            if s[i] not in hash_map[f'{count}']:
                hash_map[f'{count}'] += value
            else:
                count += 1
                hash_map[f'{count}'] = value
        return  hash_map

# max(len(i) for i in hash_map.values())       
    
Solution().lengthOfLongestSubstring(s ='dvdk')    
```
